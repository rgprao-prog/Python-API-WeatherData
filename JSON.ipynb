```python
import json

student_string = """
{
    "students": [
        {
            "name": "Jane Doe",
            "student_ID" : "1234",
            "classes" : ["Maths", "Biology", "Comp Sci"],
            "has_liscence" : true,
            "emails" : null
        },
        {
            "name": "Hermione Granger",
            "student_ID" : "5678",
            "classes" : ["Potions", "Defence", "Herbology"],
            "has_liscence" : true,
            "emails" : null
        }
    ]
}"""

print("student_string is a ", type(student_string))   #json string
data = json.loads(student_string)
print(data)
print("data is the data type", type(data))  #weve parsed the json data into python dic 
# you can see how true and false have capitalized and null --> None 
# the data in the json contains the key, student and then an array with objects, the array should be a list now as seen below
print(type(data["students"]))

# using loops:
for student in data["students"]:
    print(student["name"], student["classes"])
    del student["emails"]

#lets say we remove the emails from each student and convert this back into a json string
new_string = json.dumps(data, indent = 2)   # the indent parameter is used to format the result
print("new_string:", new_string)



```

```python
# this example uses json files and the methods dump() and load()

import json

with open("students.json") as f:
    students_parse = json.load(f)   #we are using python to parse the json file into dict

#uing indexing and keys to obtain data:
students_parse["students"][0]["courses"][2] # gives luna lovegoods 3rd course that she's taking 

for student in students_parse["students"]:      #using a for loop to collect the names of students who study a certain course
    if "Potions" in student["courses"]:
        print(f"the students that study potions are {student['name']}")


new_student = {                             #creating a new student to add to json file 
    "name": "Ronald Weesly",
            "student_id": "0005",
            "house": "Gryffindor",
            "courses": [
                "Defence Against the Dark Arts",
                "Potions",
                "Charms"
            ],
            "has_broom_permit": True,
            "owl_address": None
}

students_parse["students"].append(new_student)      #appending this new student at the CORRECT location - in this case its in the list of the students key 

with open("students.json", "w") as f:
    json.dump(students_parse, f, indent = 4)    #.dump(python_object, file_object)

    #remember w mode clears the existing file contents.
```

PYTHON REQUESTS LIBRARY:

Python's requests library makes it easy to communicate with web servers using HTTP. The main requetss methods are 
>> requests.get(url)
>> requests.delete(url)
>> requests.post(url)
>> requests.put(url)
>> requests.patch(url)

additional methods include:
>> requests.head : Similar to GET, but asks for the response headers without downloading the response body.
    Useful when you want to find things like: content type, content length, last modified date, server information --> without retrieving the entire body.
>> 






```python
import requests
import json

#url we want to send the request to:
url = 'https://jsonplaceholder.typicode.com/posts/1'    #fake url used when testing things
response = requests.get(url)       #using the requests.get() to GET info from the URL
#the response is that the server will send us is stored in the response variable

print(f"status code: {response.status_code}")   #printing out the status code of this response
print(f"Content-Type: {response.headers.get("content-type")}")  #getting the content type from the response header data (case insensitive)

data = response.json() #parses json data that came from the request into python dict 
print("this is the json data:", json.dumps(data, indent=4)) #this is just to see the data clearly

print("'Post title:'", data["title"])  #selecting and printing the title from the dict
```

```python
import requests
import json

url = 'https://reqres.in/api/users' 

params = {"page" : 2}
response = requests.get(url, params=params)
print(f"{response.url} containing the query params")

response.raise_for_status() #raises a 4xx/5xx error status if there is one and the program doesnt continue 

data = response.json()
print(json.dumps(data, indent=4))
print(data["data"][1]["email"]) #getting the email at this particular location

print("Page", data["page"])
for user in data["data"]:
    print(user["id"], user["email"])
```

```python
#ERROR HANDLING

import requests

try:
    response = requests.get("https://httpbin.org/delay/3", timeout=2)   #by default this url takes 3 seconds to respond 
    response.raise_for_status()
    print("Sucess", response.json())
except requests.exceptions.Timeout:
    print("the request timed out")

except requests.exceptions.RequestException as e:
    print("Request Failed", e)
```
